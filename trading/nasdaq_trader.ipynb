```python
import pandas_datareader as pdr 
import yfinance as yf 
import numpy as np
import pandas as pd
```

```python
et_data = pdr.nasdaq_trader.get_nasdaq_symbols()
```

```python
etfs = et_data[et_data.ETF]
etfs[etfs['NASDAQ Symbol']=='SPY']
```

```python
tickers_list = etfs[etfs['Listing Exchange']=='P']['NASDAQ Symbol'].to_list()
```

```python
#etfs.shape[0]
```

```python
batch_size=10
```

```python
start_idx = np.linspace(0, len(etfs),batch_size, dtype=int)
start_idx
```

```python
st_idx = range(0, len(etfs), )
```

```python
all_data = []
for idx, cur in enumerate(start_idx[:-1]):
    nxt = start_idx[idx+1]
    print(cur,nxt)
    data = yf.download(tickers_list[cur:nxt], start = '2024-07-01')
    all_data.append(data)
```

```python
all_data
```

```python
all_data_df = pd.concat(all_data)
```

```python
all_data_df.Close['QQQ'].pct_change().std()
```

```python
all_data_df.Close['TQQQ'].pct_change().std()
```

```python
all_data_df.Close.pct_change().std().sort_values(ascending=False).dropna()[:30]
```
